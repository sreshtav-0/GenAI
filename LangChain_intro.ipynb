```python
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
def get_weather(city:str)->str:
    """Get the weather for a city"""
    return f"The weather in {city} is sunny"

agent = create_agent(
    model = "gpt-5.5", tools=[get_weather], system_prompt="You are a helpful assistant"
)
agent
```

```python
response = agent.invoke({"messages":[{"role":"user", "content": "What is the weather like for Hyderabad?"}]})
```
